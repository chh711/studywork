Numpy数组(ndarray)的核心特性
多维性:支持0维(标量)、1维(向量)、2维(矩阵)及更高维数组。
同质性:所有元素类型必须一致(通过dtype指定)。
高效性:基于连续内存块存储，支持向量化运算。

1.虚拟环境到底解决了什么问题？如果所有项目共用一个全局环境会发生什么？
“版本冲突”和“环境污染”问题——它让每个项目拥有自己独立的“包仓库”，互不干扰。

如果所有项目共用一个全局环境，最致命的后果是：陷入永无止境的“依赖地狱”——项目之间因所需的第三方库版本不同而相互冲突

2.列表推导式和 for 循环怎么取舍？什么时候不该用推导式？
要做的事情是“把 A 列表变成 B 列表”，且逻辑简单直白用推导式
逻辑复杂，要中途 break、continue 或复杂的异常处理，列表推导式内部无法优雅地处理异常用for循环

3.*args 和 **kwargs 分别在什么场景用？举一个你自己写的例子
*args：用来接收任意数量的位置参数（多余的无人认领的位置参数），打包成元组 (tuple)

**kwargs：用来接收任意数量的关键字参数（多余的无人认领的关键字参数），打包成字典 (dict)

*args 负责“路径的深浅”（不确定长度），**kwargs 负责“参数的多少”（不确定种类）
