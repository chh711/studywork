```python
# Numpy 向量化 vs For 循环 性能对比笔记

> 测试环境：1,000,000 条模拟传感器数据（温度、湿度）

## 测试结果

| 任务 | for 循环 | numpy 向量化 | 加速比 |
|------|---------|-------------|--------|
| 条件计数 (temp>30) | 139.88 ms | 4.2094 ms | **33x** |
| 均值+标准差 | 231.15 ms | 3.2506 ms | **71x** |
| 异常值检测 | 138.28 ms | 4.2506 ms | **33x** |

## 结论

1. **numpy 向量化全面碾压 for 循环**，在百万级数据上加速比可达 **数十到数百倍**。
2. **根本原因**：for 循环在 Python 解释器中逐元素执行，每次迭代都有类型检查、对象创建等开销；
   numpy 在 C 层面批量操作连续内存数组，充分利用 CPU 缓存和 SIMD 指令。
3. **实践建议**：
   - 能用 numpy 内置函数（`np.sum`, `np.mean`, `np.std`）就不要写循环
   - 条件筛选用布尔索引（`arr[arr > 30]`）代替 `for + if`
   - 复杂逻辑优先考虑 `np.where`, `np.select`, 布尔掩码等向量化方案
   - 只有在逻辑无法向量化时，才考虑 `numba` JIT 或 `cython` 加速循环
```
